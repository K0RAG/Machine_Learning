```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('housing.csv')
df.head()
```

```python
def compute_cost(x_train,y_train,w,b):
    
    m=len(x_train)
    f=np.dot(x_train,w)+b
    error=f-y_train
    cost=np.sum(error**2)/(2*m)

    return cost
```

```python
def compute_gradient(x,y,w,b):

    m=len(x)
    f=np.dot(x,w)+b
    error=f-y
    dj_dw=np.dot(x.T,error)/m
    dj_db=np.sum(error)/m
    
    
    return dj_dw, dj_db
```

```python
def gradient_descent(x,y,w,b,alpha,epoch):
    cost_history=[]

    for i in range(epoch):
        dj_dw,dj_db=compute_gradient(x,y,w,b)
        w=w-alpha*dj_dw
        b=b-alpha*dj_db
        cost=compute_cost(x,y,w,b)
        cost_history.append(cost)
    
    return w,b,cost_history
```

```python
#dropping unnessesary columns and null values
df.drop(columns=['ocean_proximity'],inplace=True)
print(df.isnull().sum())
df['total_bedrooms'] = df['total_bedrooms'].fillna(df['total_bedrooms'].mean())
y_train=df['median_house_value']
x_train=df.drop(columns=['median_house_value'])

#using feature scaling
mean=np.mean(x_train,axis=0)
std=np.std(x_train,axis=0)
x_norm=(x_train-mean)/std
```

```python
#taking x and y to numpy to for calculation
X=x_norm.to_numpy()
Y=y_train.to_numpy()

m,n=np.shape(X)
w=np.zeros(n)
b=0
alpha=0.01
epoch=2000
w,b,cost_history=gradient_descent(X,Y,w,b,alpha,epoch)

Y_pred=np.dot(X,w)+b
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(Y, Y_pred)
print("R2 Score:", r2)
```

```python
plt.plot(cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost Convergence")
plt.show()
```
