```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('data.csv')
df
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def compute_cost(x_train,y_train,w,b):
    m=len(x_train)
    z=np.dot(x_train,w)+b
    f=sigmoid(z)
    f = np.clip(f, 1e-15, 1 - 1e-15) #to avoid log(0) case
    cost=np.sum(y_train*np.log(f)+(1-y_train)*np.log(1-f))/(-m)

    return cost
```

```python
def compute_gradient(x,y,w,b):
    m=len(x)
    z=np.dot(x,w)+b
    f=sigmoid(z)
    error=f-y
    dj_dw=np.dot(x.T,error)/m
    dj_db=np.sum(error)/m

    return dj_dw,dj_db
    
```

```python
def gradient_descent(x,y,w,b,alpha,epoch):
    cost_history=[]

    for i in range(epoch):
        dj_dw,dj_db=compute_gradient(x,y,w,b)
        w=w-alpha*dj_dw
        b=b-alpha*dj_db
        
        cost=compute_cost(x,y,w,b)
        cost_history.append(cost)
    
    return w,b,cost_history
```

```python
#mapping and dropping unnessesary columns
df.drop(columns=['id','Unnamed: 32'],inplace=True)
print(df.isnull().sum())
df['diagnosis']=df['diagnosis'].map({ 'M':1,'B':0})
x_train=df.drop(columns=['diagnosis'])
y_train=df['diagnosis']

#using feature scaling
mean=np.mean(x_train,axis=0)
std=np.std(x_train,axis=0)
x_norm=(x_train-mean)/std
```

```python
X=x_norm.to_numpy()
Y=y_train.to_numpy()

m,n=np.shape(X)
w=np.zeros(n)
b=0
alpha=0.01
epoch=2000
w,b,cost_history=gradient_descent(X,Y,w,b,alpha,epoch)

z=np.dot(X,w)+b
Y_prob=sigmoid(z)
Y_pred = (Y_prob >= 0.5).astype(int)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

print("Accuracy:", accuracy_score(Y,Y_pred))
print(confusion_matrix(Y,Y_pred))
```

```python
plt.plot(cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost Convergence")
plt.show()
```
